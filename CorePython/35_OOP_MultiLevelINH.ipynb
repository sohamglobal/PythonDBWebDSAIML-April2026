```python
class One:
  def calcsquare(self,n):
    sq=n**2
    print(f"Square of {n} is {sq}")
```

```python
# intermediate
class Two(One):
  def add(self,a,b):
    c=a+b
    print(f"sum of {a} and {b} is {c}")
```

```python
class Three(Two):
  def finddiscount(self,amount):
    discount=amount*10/100
    print(f"discount will be {discount}")
```

```python
obj=Three()
obj.add(45,78)
obj.finddiscount(7950)
obj.calcsquare(75)
```
