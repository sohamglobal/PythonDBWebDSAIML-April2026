```python
from abc import ABC, abstractmethod
```

if a function is declared as abstract, it cannot have body. the body will be written in the derived class

```python
class Payment(ABC):
  @abstractmethod   # decorator
  def pay(self):
    pass
```

```python
class UPIPayment(Payment):
  def pay(self):
    print("bill paid using PhonePe")
  def showhistory(self):
    print("list of past transactions")
```

```python
obj=UPIPayment()
obj.pay()
obj.showhistory()
```
