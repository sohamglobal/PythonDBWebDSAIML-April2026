```python
class Product:
  quantity=0
  purchcost=0
  def __init__(self,quantity,purchcost):
    self.quantity=quantity
    self.purchcost=purchcost
  def showstock(self):
    print(f"Quantity : {self.quantity}")
    print(f"Purchase Cost : {self.purchcost}")
  def __add__(self,obj):
    q=self.quantity+obj.quantity
    c=self.purchcost+obj.purchcost
    return Product(q,c)
```

```python
dell=Product(12,842100)
dell.showstock()
```

```python
hp=Product(19,1460200)
hp.showstock()
```

```python
total=dell+hp
total.showstock()
```

add,sub,mul,mod functions are available for overloading
