```python
class One:
  def calcsquare(self,n):
    sq=n**2
    print(f"Square of {n} is {sq}")
```

```python
class Two(One):
  def add(self,a,b):
    c=a+b
    print(f"sum of {a} and {b} is {c}")
```

```python
class Three(One):
  def finddiscount(self,amount):
    discount=amount*10/100
    print(f"discount will be {discount}")
```

```python
obj1=Two()
obj1.add(11,22)
obj1.calcsquare(23)
# obj1.finddiscount(1200)
```

```python
obj2=Three()
obj2.finddiscount(3450)
obj2.calcsquare(57)
```
