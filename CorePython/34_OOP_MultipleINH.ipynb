```python
class One:
  def calcsquare(self,n):
    sq=n**2
    print(f"Square of {n} is {sq}")
```

```python
class Two:
  def add(self,a,b):
    c=a+b
    print(f"sum of {a} and {b} is {c}")
```

```python
class Three(One,Two):
  def finddiscount(self,amount):
    discount=amount*10/100
    print(f"discount will be {discount}")
```

```python
obj=Three()
obj.finddiscount(12400)
obj.calcsquare(9)
obj.add(12,45)
```
