```python
db=client["spiderdb"]
wcoll=db["coworkers"]
```

```python
nm=input('Enter worker name : ')
d1={}
d1["name"]=nm
print(d1)

lo=input('Enter new location : ')
d2={}
d2['location']=lo
print(d2)

d3={}
d3['$set']=d2
print(d3)

result=wcoll.update_one(d1,d3)
print(result.modified_count)
if result.modified_count>0:
  print('Worker location changed successfully')
else:
  print('Not found..update failed')
```

```python
db2=client["sohamdb"]
ucoll=db2["users"]
```

```python
uid=input('Enter userid : ')
d1={}
d1["userid"]=uid
print(d1)

ps=input('Enter new password : ')
d2={}
d2["password"]=ps
print(d2)

d3={"$set":d2}
print(d3)

result=ucoll.update_one(d1,d3)
if result.modified_count>0:
  print('password changed successfully...')
else:
  print('user does not exist')
```

```python
nm=input('Enter worker name : ')
d1={"name":nm}
print(d1)

iv=float(input('Enter increment value : '))
d2={"salary":iv}
d3={"$inc":d2}
print(d3)

result=wcoll.update_one(d1,d3)
if result.modified_count:
  print(f"salary of {nm} incremented by {iv}")
else:
  print('salary update failed...')
```
